```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("hr_dataset.csv")
print(data)
```

```python
data.isnull().sum()
```

```python
data=data.drop_duplicates()
```

```python
data.columns=data.columns.str.strip().str.lower()
```

```python
data.dropna(inplace=True)
```

```python
data.info()
```

```python
data.columns=data.columns.str.strip().str.lower().str.replace(' ', '_')
print(data.columns)
```

```python
data.describe()
```

```python
data.nunique()
```

```python


bins = [18, 25, 35, 45, 55, 65]

labels = [
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    '56-65'
]

data['age_group'] = pd.cut(data['age'], bins=bins, labels=labels)

print(data['age_group'].head())
```

```python
data['age_group'].value_counts().sort_index()
```

```python
education_map = {
    1: 'Below College',
    2: 'College',
    3: 'Bachelor',
    4: 'Master',
    5: 'Doctor'
}

data['education_level'] = data['education'].map(education_map)
print(data[['education', 'education_level']].head())
```

```python
jobsatistfaction_level ={
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Very High'
}
data['jobsatisfaction_level'] = data['jobsatisfaction'].map(satistfaction_level)
```

```python
environment_satisfaction_level = {
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Very High'
}
data['environment_satisfaction_level'] = data['environmentsatisfaction'].map(environment_satisfaction_level)
```

```python
worklife_balance_level = {
    1: 'Bad',
    2: 'Good',
    3: 'Better',
    4: 'Best'
}
data['worklife_balance_level']=data['worklifebalance'].map(worklife_balance_level)
```

```python
jobinvalment_level = {
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Very High'
}
data['jobinvalment_level'] = data['jobinvolvement'].map(jobinvalment_level)
```

```python
performance_level = {
    1: 'Low',
    2: 'Good',
    3: 'Excellent',
    4: 'Outstanding'
}
data['performance_level'] = data['performancerating'].map(performance_level)
```

```python
print(data.head())
```

```python
data.columns
```

```python
relationships_satisfaction_level = {
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Very High'
}
data['relationships_satisfaction_level'] = data['relationshipsatisfaction'].map(relationships_satisfaction_level)
```

```python
data['attrition_flag'] = data['attrition'].map({'Yes':1,'No':0})
```

```python
# Employee Attrition rate

attrition_rate = (
    data['attrition'].value_counts(normalize=True) * 100
)

print(attrition_rate)
```

```python
dept_attrition = pd.crosstab(
    data['department'],
    data['attrition']
)
print(dept_attrition)
```

```python
data.groupby('department')['monthlyincome'].mean()
```

```python
pd.crosstab(
    data['overtime'],
    data['attrition'],
    normalize='index'
) * 100
```
